## Projektidee

In diesem Projekt werden Zusammenhänge zwischen Stress,
Schlafdauer, Schlafqualität und verschiedenen Berufsgruppen
anhand des Sleep Health and Lifestyle Dataset untersucht.

## Fragestellungen

### 1. Stress und Schlafqualität
In welchem Zusammenhang stehen das Stressniveau und
die Schlafdauer mit der Schlafqualität?

Dazu werden folgende Unterfragen untersucht:

1. Besteht ein Zusammenhang zwischen dem Stressniveau und der Schlafqualität?
2. Wie hängt die Schlafdauer mit der Schlafqualität zusammen?
3. Unterscheidet sich die durchschnittliche Schlafdauer zwischen Personen mit unterschiedlichem Stressniveau?

### 2. Berufsgruppen und Schlafgesundheit
Inwiefern unterscheiden sich Schlafdauer, Schlafqualität
und Stressniveau zwischen verschiedenen Berufsgruppen?

Dazu werden folgende Unterfragen untersucht:

1. Welche Berufsgruppen weisen die höchste beziehungsweise niedrigste durchschnittliche Schlafqualität auf?
2. Gibt es Unterschiede in der durchschnittlichen Schlafdauer zwischen den Berufsgruppen?
3. Welche Berufsgruppen weisen das höchste durchschnittliche Stressniveau auf?
